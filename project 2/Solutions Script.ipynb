### ProblemC5: Close connections

```python
# call deque and default dictionary
from collections import defaultdict, deque

#import numpy as np

# define a function
def count_close_connection(n, m, s, k):
    """ count the number of reacheable vertices provided number of k steps from point s"""
    visited = [False] * (n + 1)
    Queue = deque([(s, 0)])
    visited[s] = True
    points_count = 0

    while Queue:
        vertex, steps = Queue.popleft()
        points_count += 1
        if steps == k:
            continue
        # traversing neighbours
        for neighbour in m[vertex]:
            if not visited[neighbour]:
                visited[neighbour] = True
                Queue.append((neighbour, steps + 1))
    return points_count

def close_connection(test_sets):
    """ Identifying the close connection to s"""
    results = []
    for case in test_sets:
        n, m, s, k = case["params"]
        edges = case["edges"]
        topology = defaultdict(list)  

        for a, b in edges:
            topology[a].append(b)

        results.append(count_close_connection(n, topology, s, k))
        print(results)
    return results

# running the close_connection function
trials_count = int(input("Enter the number test scenarios: "))
trials =[]
for i in range(trials_count):
    n, m, s, k = map(int, input("Enter the parameters n, m, s, k").split())
    edges = [tuple(map(int, input().split())) for i in range(m)]
    trials.append({"params": (n,m,s,k), "edges":edges})
results = close_connection(trials)
for result in results:
    print(result)
```

### ProblemC7:Even-Odd

```python
def even_odd(array):
    """
    Optimized version of calculating violations with a single traversal.
    """
    max_value = max(array)  # To optimize range-based checks
    
    # Frequency array to store counts of even numbers seen so far
    freq_even = [0] * (max_value + 1)  
    
    violation_count = 0
    
    for num in array:
        if num % 2 == 1:  # Odd number
            # Count all even numbers less than current odd number
            violation_count += sum(freq_even[:num])
        else:  # Even number
            # Update frequency of the current even number
            freq_even[num] += 1
    
    return violation_count

# Input handling
test_count = int(input("Enter the number of test scenarios: "))
results = []

for _ in range(test_count):
    n = int(input("Enter size: "))
    array = list(map(int, input("Enter the array: ").split()))
    results.append(even_odd(array))

# Output the results
for result in results:
    print(result)  
```

```python
def even_odd(array):
    def merge_sort(array, left, right):
        if left >= right:
            return 0

        mid = (left + right) // 2

        violation = merge_sort(array, left, mid)
        violation += merge_sort(array, mid + 1, right)
        violation += merge(array, left, mid, right)

        return violation

    def merge(array, left, mid, right):
        B = []  # Temporary array for merging
        current_left = left
        current_right = mid + 1
        my_even_count = 0
        sub_violation = 0

        while current_left <= mid and current_right <= right:
            if array[current_left] > array[current_right]:
                if array[current_right] % 2 != 0:
                    sub_violation += my_even_count
                B.append(array[current_right])
                current_right += 1
            else:
                if array[current_left] % 2 == 0:
                    my_even_count += 1
                B.append(array[current_left])
                current_left += 1

        while current_left <= mid:
            if array[current_left] % 2 == 0:
                my_even_count += 1
            B.append(array[current_left])
            current_left += 1

        while current_right <= right:
            if array[current_right] % 2 != 0:
                sub_violation += my_even_count
            B.append(array[current_right])
            current_right += 1

        # Overwrite the original array with merged results
        for i in range(len(B)):
            array[left + i] = B[i]

        return sub_violation

    return merge_sort(array, 0, len(array) - 1)


def even_odd_violations(test_cases):
    results = []
    for array in test_cases:
        count = even_odd(array)
        results.append(count)
    return results


# Input Handling
test_count = int(input("Enter the number of test scenarios: "))
test_cases = []

for _ in range(test_count):
    n = int(input("Enter input size: "))  # Can be skipped, as array length is inferred.
    array = list(map(int, input("Enter the array: ").split()))
    test_cases.append(array)

# Processing and Output
outputs = even_odd_violations(test_cases)
for result in outputs:
    print(result)
```
