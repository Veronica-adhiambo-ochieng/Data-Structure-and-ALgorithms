# Part 2: Introduction to pandas
**pandas** is an open-source Python library for working with structured data. It provides two main data structures:
- **Series**: a one-dimensional labeled array holding data of any type (integers, strings, Python objects, ...).
- **DataFrame**: a two-dimensional table with rows and columns, where columns can have different types.

With pandas you can load data from CSV or Excel files, clean and transform it, and explore it with a few lines of code.

````python
import pandas as pd
import numpy as np
````

## 1. [Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html)
A Series is created with:

`s = pd.Series(data=..., index=..., name=...)`

where `data` can be a Python list, a dict, a NumPy array or a scalar value.

Specify only `data`: pandas creates a default integer index.

````python
s = pd.Series([2, 4, 5])
s
````

Specify `index` to use your own labels, and use them to access or modify elements.

````python
s = pd.Series([2, 4, 5], index=['A', 'B', 'C'])
s
````

````python
s['A']
````

````python
s['B'] = 10
s
````

Specify `name` to give the Series a meaningful name.

````python
s = pd.Series([2, 4, 5], index=['A', 'B', 'C'], name='my_series')
s
````

If `data` is a dictionary, its keys automatically become the index labels. `.values` returns the underlying values as a NumPy array.

````python
d = {'A': 2, 'B': 4, 'C': 5}
s = pd.Series(d)
print(s)
print(s.values)
````

### Series and NumPy
A `Series` is built on top of a NumPy array and supports vectorised math and broadcasting. The big difference is that pandas keeps the **index labels** and **aligns by index** when combining Series.

`NaN` (not a number) is the standard marker for missing data in pandas. In code, it is written as `np.nan`.

````python
s = pd.Series([10, 20, np.nan, 40], index=["a", "b", "c", "d"])
print(s)
print(s.isna())
````

````python
a = s * 2 + 5                          # scalar broadcasting: [25, 45, NaN, 85]
b = np.where(s.isna(), 'Missing', s)   # elementwise selection with NumPy (returns a NumPy array)

print(a)
print(b)
````

When adding two Series, pandas aligns them **by index label**. Labels that are missing in either Series give `NaN`.

````python
t = pd.Series([1, 2, 3], index=["b", "c", "e"])

print(s + t)
````

NumPy arrays have no labels, so they are combined **by position**. Arrays of different lengths cannot be added:

````python
s_np = s.to_numpy()
t_np = t.to_numpy()
print(s_np.shape, t_np.shape)

# s_np + t_np  # uncomment to see the error: shapes (4,) and (3,) do not match
````

## 2. [DataFrames](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html)
A DataFrame holds columns of potentially different types. You can think of it as a spreadsheet, or as a dict of Series. It is the most commonly used pandas object.

`df = pd.DataFrame(data=..., index=..., columns=...)`

where `data` can be a dict of 1D arrays, lists, dicts or Series, a 2D NumPy array, a Series, or another DataFrame.

````python
df = pd.DataFrame([2, 4, 5])
print(df.shape)
df
````

Create a DataFrame from a dictionary of Series. The index labels are aligned, and missing entries are filled with `NaN`.

````python
d = {
    "one": pd.Series([1.0, 2.0, 3.0], index=["a", "b", "c"]),
    "two": pd.Series([1.0, 2.0, 3.0, 4.0], index=["a", "b", "c", "d"]),
}

df = pd.DataFrame(d)
df
````

## 3. Indexing: `[]`, `.loc`, `.iloc`

- `s[...]` / `df[...]` (**square brackets**):
  - `Series`: label-based selection, boolean filtering, or label slices (endpoints inclusive). Beware: if the index consists of integers, `s[1]` means label 1, not the second element.
  - `DataFrame`: select columns by label (or list of labels), or filter rows with a boolean mask. Slices like `df[0:3]` select rows by position.
- `.loc[row_selector, col_selector]`: label-based (endpoints inclusive for slices).
- `.iloc[row_selector, col_selector]`: position-based (endpoints exclusive for slices).

````python
s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])

df = pd.DataFrame(
    {"x": [10, 20, 30, 40],
     "y": [5,  6,  7,  8]},
    index=["a", "b", "c", "d"]
)
````

### Series: `[]`, `.loc`, `.iloc`

````python
print(f'Label:\n{s["b"]}\n')                   # 20
print(f'List of labels:\n{s[["b", "d"]]}\n')   # b 20, d 40
print(f'Label slice:\n{s["b":"d"]}\n')         # inclusive: b 20, c 30, d 40
print(f'Boolean filter:\n{s[s > 15]}')         # b 20, c 30, d 40
````

````python
print(s.loc["b"])   # label-based: same as s["b"]
print(s.iloc[1])    # position-based: second element
````

### DataFrame: `[]`

````python
print(df["x"], "\n")            # a Series: the 'x' column
print(df[["x", "y"]], "\n")     # a DataFrame with two columns
print(df[df["x"] > 15], "\n")   # boolean filter on rows
print(df[0:2])                  # row slice by position: rows a, b
````

### DataFrame: `.loc` (labels) and `.iloc` (positions)

````python
a = df.loc[['c', 'd'], ['x']]  # rows c, d and column x
print(type(a))
a
````

````python
a = df.iloc[1, 0]   # second row, first column
print(a)
````

### 3.4 Map values
`df['col'] = df['col'].map({old_value_1: new_value_1, old_value_2: new_value_2})`

You can overwrite the column or create a new one. Creating a new one lets you check the mapping first; afterwards, remove the old column with [`drop`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop.html). Note that with a dict, values that are not in the mapping become `NaN`.

In this dataset, `num` (now `target`) is 0 for patients without heart disease and 1 to 4 for patients with heart disease (increasing severity). We map it to a binary diagnosis:

````python
df_renamed["diagnosis"] = df_renamed["target"].map({0: "no", 1: "yes", 2: "yes", 3: "yes", 4: "yes"})
df_renamed[["target", "diagnosis"]].head()
````

````python
df_renamed = df_renamed.drop(columns='target')
````

### `map` and `apply`
- `Series.map`: elementwise on a Series; remaps values with a dict/Series or applies a simple function.
- `Series.apply`: elementwise on a Series with a function; useful for more complex logic returning one value per element.
- `DataFrame.apply`: applies a function per column (`axis=0`) or per row (`axis=1`); the function receives a whole Series (a column or a row).
- `DataFrame.map` (formerly `applymap`): elementwise on every cell of a DataFrame.

**`Series.map`** for simple logic:

````python
# Convert age in years to age in months
df_renamed["age months"] = df_renamed["age"].map(lambda x: x * 12)
````

**`Series.apply`** for more complex logic:

````python
# Bin a numeric feature with custom rules
def trestbps_bin(x):
    if x < 115:
        return "low"
    elif x < 130:
        return "medium"
    else:
        return "high"

df_renamed["trestbps_cat"] = df_renamed["trestbps"].apply(trestbps_bin)
df_renamed[["trestbps", "trestbps_cat"]].head()
````

**`DataFrame.apply`** by row (`axis=1`) or by column (`axis=0`):

````python
# By row: the function receives one row at a time
def chol_trestbps_ratio(row):
    return row["chol"] / row["trestbps"]

df_renamed["chol_trestbps_ratio"] = df_renamed[["chol", "trestbps"]].apply(chol_trestbps_ratio, axis=1)

# By column: standardise chol and trestbps
features = ["chol", "trestbps"]
standardised = df_renamed[features].apply(lambda col: (col - col.mean()) / col.std(), axis=0)
standardised.head()
````
