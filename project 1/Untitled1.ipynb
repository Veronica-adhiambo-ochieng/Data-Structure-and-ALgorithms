# Binary search

```python
def binary_search_Iterative(array, left, right, x):
    """
    Binary Search Algorithm is a searching algorithm used in a sorted array by repeatedly dividing the 
    search interval in half. The idea of binary search is to use the information that the array is sorted
    and reduce the time complexity to O(log N). 
    Args:
    A (list): A sorted list of integers (e.g., [3, 4, 5, 7, 12, 14, 16, 18, 19, 32, 43, 77]).
    x (int): The target integer to search for in `A` (e.g., 21).
  """
    while left <= right:
        mid = (left + right)//2
        # check if the middle value is equal to x, then we stop
        if array[mid] == x:
            return x
        # if x is greater than mid, ignore left half (array) and split the next array
        if array[mid] < x:
            left = mid + 1
        # if x is less than mid, ignore right half (array) and work with values below
        else:
            right = mid -1
    # element is the present
    return -1
```

```python
def binary_search_recursive(array, left, right, x):
    """
    Binary Search Algorithm is a searching algorithm used in a sorted array by repeatedly dividing the 
    search interval in half. The idea of binary search is to use the information that the array is sorted
    and reduce the time complexity to O(log N). 
    Args:
    A (list): A sorted list of integers (e.g., [3, 4, 5, 7, 12, 14, 16, 18, 19, 32, 43, 77]).
    x (int): The target integer to search for in `A` (e.g., 21).
  """
    if (left <= right):
        mid = (left + right)//2
        # check if the value is equal to mid 
        if array[mid] == x:
            return x
        elif array[mid] > x:
            return binary_search_recursive(array, left, mid - 1, x)
        elif array[mid] < x:
            return binary_search_recursive(array, mid + 1, right, x)
    else:
        return -1
```
