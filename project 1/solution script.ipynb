### Problem4: AIMS frequency

```python
def max_aims_frequency(test_cases):
    results = []

    for n, s in test_cases:
        # Transform string into array of 1 and -1
        aims_set = {'a', 'i', 'm', 's'}
        transformed = [1 if char in aims_set else -1 for char in s]

        # Find the maximum subarray sum
        max_sum = 0
        current_sum = 0

        for value in transformed:
            current_sum = max(value, current_sum + value)
            max_sum = max(max_sum, current_sum)

        results.append(max_sum)

    return results


# Input reading
def main():
    # Get inputs
    T = int(input("Enter number of test cases: "))  # Number of test cases
    test_cases = []

    for _ in range(T):
        n = int(input("Enter the length of the string: "))  # Length of the string
        s = input("Enter the string: ")  # The string
        test_cases.append((n, s))

    # Solve the problem
    results = max_aims_frequency(test_cases)

    # Output results
    for result in results:
        print(result)


# Run
if __name__ == "__main__":
    main()
```

### Problem5: Inverting entropy

```python
import math

def binary_entropy_function_inverse(alpha):
    if alpha == 0:
        return 0.0
    if alpha == 1:
        return 0.5

    # Define the binary entropy function
    def entropy(p):
        if p == 0 or p == 1:
            return 0
        return -p * math.log2(p) - (1 - p) * math.log2(1 - p)

    # Binary search
    low, high = 0, 0.5
    while high - low > 1e-7:  # Precision >= 10^-6
        mid = (low + high) / 2
        if entropy(mid) < alpha:
            low = mid
        else:
            high = mid

    return (low + high) / 2  # Return the midpoint


# Input reading and processing
def main():
    T = int(input("Enter the number of test sets: "))  # Number of test cases
    results = []
    for _ in range(T):
        alpha = float(input("Enter alpha (0 <= alpha <= 1): "))
        results.append(binary_entropy_function_inverse(alpha))

    # Print results
    for result in results:
        print(f"{result:.10f}")


if __name__ == "__main__":
    main()
```

```python
def mars(day, month, year):
    # Function to compute the sols for months before the given month
    def months_before(month):
        return 27 * ((month - 1) // 6) + 28 * ((month - 1) % 6)

    # Function to compute the sols for years before the given year
    def years_before(year, base_year):
        diff = year - base_year
        leap_years = diff // 2 + diff % 2 + diff // 10
        return 669 * leap_years + 668 * (diff - leap_years)

    # Calculate the base sols (for reference date 220.24.28)
    base_sols = 28 + months_before(24) + years_before(220, 1)

    # Precompute the sols for the given month and year only once
    months_sols = months_before(month)
    if year > 220:
        # If the year is after the reference year
        total_sols = day + months_sols + years_before(year, 221)
    else:
        # If the year is before or equal to the reference year
        total_sols = base_sols - (day + months_sols + years_before(year, 1))

    # Compute the day of the week (0 = Sunday, 1 = Monday, ..., 6 = Saturday)
    return (total_sols + 1) % 7 if year > 220 else (1 - total_sols) % 7
```
